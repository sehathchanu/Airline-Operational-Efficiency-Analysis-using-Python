```python
import pandas as pd

# Load dataset
df = pd.read_csv("airlines.csv")

# Basic info
print(df.head())
print(df.info())
print(df.describe())
```

```python
# Check missing values
print(df.isnull().sum())

# Drop or fill missing values
df = df.dropna()   # simple method

# Remove duplicates
df = df.drop_duplicates()
```

```python
df['Time.Label'] = pd.to_datetime(df['Time.Label'])
```

```python
df['Date'] = pd.to_datetime(df['Time.Year'].astype(str) + '-' + df['Time.Month'].astype(str) + '-01')
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.histplot(df['Statistics.Minutes Delayed.Total'], bins=30, kde=True)
plt.title("Total Delay Minutes Distribution")
plt.show()
```

```python
df[['Statistics.Flights.On Time', 'Statistics.Flights.Delayed']].sum().plot(kind='bar')

import matplotlib.pyplot as plt
plt.title("On-Time vs Delayed Flights")
plt.show()
```

```python
delay_cols = [
    'Statistics_Minutes_Delayed_Carrier',
    'Statistics_Minutes_Delayed_Weather',
    'Statistics_Minutes_Delayed_Late_Aircraft',
    'Statistics_Minutes_Delayed_Security',
    'Statistics_Minutes_Delayed_National_Aviation_System'
]

df[delay_cols].sum().plot(kind='bar')

plt.title("Delay Causes Analysis")
plt.ylabel("Total Delay Minutes")
plt.xticks(rotation=45)
plt.show()
```

```python
sns.lineplot(x='Time_Label', y='Statistics_Minutes_Delayed_Total', data=df)

plt.title("Total Delay Over Time")
plt.xticks(rotation=45)
plt.show()
```

```python
sns.barplot(x='Time_Month_Name', y='Statistics_Minutes_Delayed_Total', data=df)

plt.title("Monthly Delay Analysis")
plt.xticks(rotation=45)
plt.show()
```

```python
sns.scatterplot(x='Statistics_Flights_Total', y='Statistics_Minutes_Delayed_Total', data=df)

plt.title("Flights vs Total Delay")
plt.show()
```
